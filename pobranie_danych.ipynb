```python
import requests
import pandas as pd

# Wklej tutaj skopiowany link z DevTools
url = 'https://wyniki.datasport.pl/results5969/results.json'

# Pobranie danych bezpośrednio z API Datasportu
response = requests.get(url)
dane = response.json()

# Wrócenie danych do struktury DataFrame
df = pd.DataFrame(dane)

# Opcjonalny podgląd, żeby sprawdzić czy weszły dobre kolumny
print(df.head())

# Zapis do pliku CSV (index=False usuwa pierwszą kolumnę z numeracją wierszy Pandas)
df.to_csv('biegnij_warszawo_2026_surowe.csv', index=False)
```
