# WCZYTANIE DANYCH I INFORMACJE O ZBIORZE

```python
import requests
import pandas as pd
import os
```

```python
def fetch_and_clean_run_data():
    # Json z narzedzi developerskich z datasportu
    url = 'https://wyniki.datasport.pl/results5969/results.json'

    print("Rozpoczynam pobieranie danych z API...")
    response = requests.get(url)

    if response.status_code == 200:
        data_json = response.json()
        print(f"Pobrano pomyślnie. Liczba wczytanych rekordów: {len(data_json)}")

        # Transformacja do df
        df = pd.DataFrame(data_json)

        # Zapisujemy cały zbiór do folderu data
        # ---WERSJA DLA COLAB---
        project_root = '/content'
        output_dir = os.path.join(project_root, 'data')
        os.makedirs(output_dir, exist_ok=True)
        # ---WERSJA DLA PYCHARM---
        #script_dir = os.path.dirname(os.path.abspath(__file__))
        #project_root = os.path.dirname(script_dir)

        output_file = os.path.join(project_root, 'data', 'dane_surowe_biegnij_warszawo.csv')
        df.to_csv(output_file, index=False, encoding='utf-8')
        print(f"Zapisano pełny zbiór do: {output_file}")

    else:
        print(f"Błąd pobierania: {response.status_code}")

if __name__ == "__main__":
    fetch_and_clean_run_data()
```

# BRAKUJĄCE DANE I USUWANIE NIEPOTRZEBNYCH KOLUMN

```python
# Analiza braków
missing_data = pd.DataFrame({
    'Liczba braków': df.isnull().sum(),
    'Procent braków (%)': (df.isnull().sum() / len(df) * 100).round(2)
})
# Sortowanie według liczby braków
missing_data = missing_data.sort_values(
    by='Liczba braków',
    ascending=True
)
display(missing_data)
```

```python
# Puste kolumny
empty_columns = df.columns[df.isnull().all()].tolist()
print("Liczba pustych kolumn:", len(empty_columns))
print("Puste kolumny:")
print(empty_columns)

# WNIOSKI
# W zbiorze znajduje się wiele pustych kolumn - zostaną one usunięte ze zbioru przed analizą
# Liczba pustych kolumn: 23
# Puste kolumny: ['nazdod2', 'nazdod3', 'nazdod4', 'nazdod5', 'nazdod6', 'nazdod7', 'nazdod8', 'uwagi', 'czasd1', 'czasd2', 'czasd3', 'czast1', 'czast2', 'nazdodw2', 'nazdodw3', 'nazdodw4', 'nazdodw5', 'nazdodw6', 'nazdodw7', 'nazdodw8', 'dns', 'dnf', 'korona']
```
